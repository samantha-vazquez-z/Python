```python
#importar librerías
from mpl_toolkits.mplot3d import axes3d
import matplotlib.pyplot as plt
from matplotlib import style
import numpy as np
import csv
import pandas as pd
```

```python
# Cargamos los datos
df = pd.read_csv('C:/Users/Vazsa/OneDrive/Documentos/Canal 1.csv')
df3 = pd.read_csv('C:/Users/Vazsa/OneDrive/Documentos/canal 3.csv')
```

```python
# Para el gráfico de disperción
fig = plt.figure()

ax1 = fig.add_subplot(111, projection='3d')

ax1.scatter(df['XM'], df['YM'], df['ZM'])
ax1.scatter(df3['XM'], df3['YM'], df3['ZM'])
ax1.legend('Canal 1')
ax1.legend('Canal 3')

ax1.set_xlabel('XM')
ax1.set_ylabel('yM')
ax1.set_zlabel('zM')

plt.title('gráfico de disperción')
plt.show()
```

```python
# para los boxplots xd
x = np.mean(df['XM'])
fig, axs = plt.subplots(3, 2, figsize=(7, 7))
axs[0, 0].boxplot(df['XM'])
axs[0, 0].set_title('Canal 1')
axs[0, 0].set_xlabel('XM')

axs[1, 0].boxplot(df['YM'])
axs[1, 0].set_xlabel('YM')

axs[2, 0].boxplot(df['ZM'])
axs[2, 0].set_xlabel('ZM')

axs[0, 1].boxplot(df3['XM'])
axs[0, 1].set_title('Canal 2')
axs[0, 1].set_xlabel('XM')

axs[1, 1].boxplot(df3['YM'])
axs[1, 1].set_xlabel('YM')

axs[2, 1].boxplot(df3['ZM'])
axs[2, 1].set_xlabel('ZM')

plt.legend()
plt.tight_layout()
plt.show()
```
